Simple price prediction model using sklearn and linear regression.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

diameters = np.array([[8], [10], [12]])
prices = np.array([10, 13, 16])

print("Hello world")
model = LinearRegression()

model.fit(diameters, prices)

new_pizza = np.array([[20]])
prediction = model.predict(new_pizza)

print(f"A 20-inch pizza is predicted to cost: ${prediction[0]:.2f}")
```
